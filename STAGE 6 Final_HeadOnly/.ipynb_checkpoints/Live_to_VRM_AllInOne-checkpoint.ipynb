# VTuber: HOG-SVM Expressions and MediaPipe Head Rotation

The hybrid demonstration uses the trained HOG-SVM classifiers for eyes and mouth and MediaPipe Face Landmarker for pitch, yaw and roll. The body remains fixed.

### live_state.mjs

```python
# Browser source
WEB_SOURCE = r'''// Latest-state transport, not a video timeline
import {expressionTargets,EXPRESSION_NAMES} from './timeline_viseme.mjs';
export {EXPRESSION_NAMES};
export function usablePacket(packet,elapsedMs){
  return Boolean(packet&&packet.active&&Number.isFinite(packet.age_ms)&&packet.age_ms>=0&&
    Number.isFinite(elapsedMs)&&elapsedMs>=0&&packet.age_ms+elapsedMs<800&&packet.row);
}
export function liveTargets(packet,elapsedMs,swap,gain){
  if(!usablePacket(packet,elapsedMs))return Array(7).fill(0);
  const row=packet.row;
  if(packet.backend==='mediapipe'&&row.continuous&&row.head_tracking_ok){
    const clamp=v=>Number.isFinite(v)?Math.max(0,Math.min(1,v)):0;
    const c=row.continuous;
    // Screen-relative sides use the same mapping as the original player
    return [clamp(swap?c.left_closed:c.right_closed),clamp(swap?c.right_closed:c.left_closed),
      clamp(c.mouth_open)*clamp(gain),0,0,0,0];
  }
  return expressionTargets(row,swap,gain);
}
'''
(LIVE_DIR / 'live_state.mjs').write_text(WEB_SOURCE, encoding='utf-8')
```

### timeline_viseme.mjs

```python
# Browser source
WEB_SOURCE = r'''// Eyes remain binary; mouth labels have six classes.
export const EXPRESSION_NAMES = ['blinkLeft', 'blinkRight', 'aa', 'ih', 'ou', 'ee', 'oh'];
export const MOUTH_TO_VRM = Object.freeze({closed:null, a:'aa', i:'ih', u:'ou', e:'ee', o:'oh'});

// CSV/JSON
// Validate ordering and labels without modifying the raw predictions.
export function validateTimeline(data) {
  if (!data || ![1, 2].includes(data.schema_version)
      || !Number.isFinite(data.fps) || data.fps <= 0
      || !Array.isArray(data.rows) || !data.rows.length)
    throw new Error(' expressions.json');
  let previous = -1;
  for (const row of data.rows) {
    if (!row || !Number.isFinite(row.time_s) || row.time_s < 0 || row.time_s <= previous)
      throw new Error('');
    previous = row.time_s;
    for (const name of ['screen_left_eye', 'screen_right_eye'])
      if (!['open', 'closed', 'unknown'].includes(row[name]))
        throw new Error('Unknown eye state: ' + name);
    const mouthLabels = [...Object.keys(MOUTH_TO_VRM), 'unknown'];
    // schema=1openaaopen。
    // Support old binary timelines as aa-only playback for compatibility.
    if (data.schema_version === 1) mouthLabels.push('open');
    if (!mouthLabels.includes(row.mouth)) throw new Error('Mouth: ' + row.mouth);
  }
  return data;
}

// Framenull。
// Use the source video's time as the master clock and binary-search its row.
export function findRow(data, time) {
  if (!data || !Number.isFinite(time) || time < 0) return null;
  const rows = data.rows;
  if (time < rows[0].time_s || time >= rows.at(-1).time_s + 1 / data.fps) return null;
  let low = 0, high = rows.length - 1;
  while (low < high) {
    const mid = Math.ceil((low + high) / 2);
    if (rows[mid].time_s <= time) low = mid;
    else high = mid - 1;
  }
  return rows[low];
}

// 7EXPRESSION_NAMES
// Build all seven target weights from zero to clear the previous mouth shape.
export function expressionTargets(row, swap = false, mouthGain = 0.7) {
  const targets = Array(EXPRESSION_NAMES.length).fill(0);
  if (!row || row.head_tracking_ok === false) return targets;
  const usable = name => row[name + '_crop_valid'] !== false;
  const leftSource = swap ? 'screen_left_eye' : 'screen_right_eye';
  const rightSource = swap ? 'screen_right_eye' : 'screen_left_eye';
  targets[0] = usable(leftSource) && row[leftSource] === 'closed' ? 1 : 0;
  targets[1] = usable(rightSource) && row[rightSource] === 'closed' ? 1 : 0;
  const gain = Number.isFinite(mouthGain) ? Math.min(1, Math.max(0, mouthGain)) : 0;
  const expression = row.mouth === 'open' ? 'aa' : MOUTH_TO_VRM[row.mouth];
  const index = EXPRESSION_NAMES.indexOf(expression);
  if (usable('mouth') && index >= 2) targets[index] = gain;
  return targets;
}
'''
(LIVE_DIR / 'timeline_viseme.mjs').write_text(WEB_SOURCE, encoding='utf-8')
```

## 3. Tracking, alignment and classification


```python
def select_roi(frame, title):
    """Select an ROI and validate its size."""

    x, y, w, h = cv2.selectROI(
        title, frame,
        showCrosshair=True,
        fromCenter=False
    )
    cv2.destroyWindow(title)

    if w < 8 or h < 8:
        raise ValueError("ROI too small or selection cancelled。")

    return int(x), int(y), int(w), int(h)

def rect_points(rect):
    """Convert a rectangle to four corner points."""
    x, y, w, h = rect

    return np.float32([
        [x, y],
        [x + w - 1, y],
        [x + w - 1, y + h - 1],
        [x, y + h - 1]
    ]).reshape(-1, 1, 2)

def track_points(previous_gray, current_gray, points, reference):
    """LKFilter LK tracks with a forward-backward check.

    referenceFrameFrame
    Keep initial coordinates to estimate the initial-to-current transform.
    """

    current, status, _ = cv2.calcOpticalFlowPyrLK(
        previous_gray, current_gray,
        points, None, **LK_PARAMS
    )

    if current is None or status is None:
        return None, None

    valid = (
        (status.ravel() == 1)
        & np.isfinite(current.reshape(-1, 2)).all(axis=1)
    )

    if not valid.any():
        return None, None

    old = points[valid]
    current = current[valid]
    reference = reference[valid]

    backward, status_back, _ = cv2.calcOpticalFlowPyrLK(
        current_gray, previous_gray,
        current, None, **LK_PARAMS
    )

    if backward is None or status_back is None:
        return None, None

    error = np.linalg.norm(
        old.reshape(-1, 2) - backward.reshape(-1, 2),
        axis=1
    )

    height, width = current_gray.shape
    xy = current.reshape(-1, 2)

    reliable = (
        (status_back.ravel() == 1)
        & np.isfinite(error)
        & (error < FB_THRESHOLD)
        & (xy[:, 0] >= 0)
        & (xy[:, 0] < width)
        & (xy[:, 1] >= 0)
        & (xy[:, 1] < height)
    )

    return current[reliable], reference[reliable]

def replenish(gray, points, reference, transform, corners):
    """Replenish corners inside the tracked head.

    
    Map new points back to reference coordinates; corners are not landmarks.
    """

    if len(points) >= REPLENISH_THRESHOLD:
        return points, reference

    mask = np.zeros(gray.shape, dtype=np.uint8)

    cv2.fillConvexPoly(
        mask,
        np.rint(corners).astype(np.int32).reshape(-1, 2),
        255
    )

    mask = cv2.erode(
        mask, np.ones((9, 9), dtype=np.uint8)
    )

    for point in points.reshape(-1, 2):
        cv2.circle(
            mask,
            tuple(np.rint(point).astype(int)),
            10, 0, -1
        )

    params = FEATURE_PARAMS.copy()
    params["maxCorners"] = (
        FEATURE_PARAMS["maxCorners"] - len(points)
    )

    if params["maxCorners"] <= 0:
        return points, reference

    new_points = cv2.goodFeaturesToTrack(
        gray, mask=mask, **params
    )

    if new_points is None:
        return points, reference

    new_reference = cv2.perspectiveTransform(
        new_points, np.linalg.inv(transform)
    )

    valid = np.isfinite(
        new_reference.reshape(-1, 2)
    ).all(axis=1)

    return (
        np.concatenate([points, new_points[valid]], axis=0),
        np.concatenate([reference, new_reference[valid]], axis=0)
    )

def locate_eye(aligned, valid_mask, eye):
    """SSDMatch an eye template with head-motion fallback."""

    x, y, w, h = eye["rect"]
    height, width = aligned.shape

    mx = max(4, int(w * SEARCH_MARGIN))
    my = max(4, int(h * SEARCH_MARGIN))

    x0, y0 = max(0, x - mx), max(0, y - my)
    x1 = min(width, x + w + mx)
    y1 = min(height, y + h + my)

    search = aligned[y0:y1, x0:x1]
    rect = eye["rect"]
    score = float("inf")
    matched = False

    if search.shape[0] >= h and search.shape[1] >= w:
        response = cv2.matchTemplate(
            search,
            eye["template"],
            cv2.TM_SQDIFF_NORMED
        )

        valid_search = (
            valid_mask[y0:y1, x0:x1] > 250
        ).astype(np.float32)

        coverage = cv2.matchTemplate(
            valid_search,
            np.ones((h, w), dtype=np.float32),
            cv2.TM_CCORR
        ) / float(w * h)

        response[coverage < 0.999] = np.inf
        response[~np.isfinite(response)] = np.inf

        if np.isfinite(response).any():
            score, _, location, _ = cv2.minMaxLoc(response)

            if score <= MATCH_THRESHOLD:
                rect = (
                    x0 + location[0],
                    y0 + location[1],
                    w, h
                )
                matched = True

    return rect, float(score), matched


def load_expression_model(path, size_key):
    """HOGLoad the classifier, HOG settings and labels.

    open/closedclosed/a/i/u/e/o。
    Eyes use two classes; the mouth uses six visual mouth-shape classes.
    """
    with open(path, "rb") as file:
        bundle = pickle.load(file)

    expected = (
        {"open", "closed"} if size_key == "eye_size"
        else {"closed", "a", "i", "u", "e", "o"}
        if size_key == "mouth_size" else None
    )
    if expected is None:
        raise ValueError(f"Unknown size key: {size_key}")

    # 
    # Use the saved label IDs rather than assuming an ID order.
    mapping = bundle["label_to_id"]
    labels = {int(value): key for key, value in mapping.items()}
    if set(mapping) != expected or len(labels) != len(mapping):
        raise ValueError(f"{path.name} {mapping}")

    model = bundle["model"]
    if set(int(value) for value in model.classes_) != set(labels):
        raise ValueError(f"{path}")
    size = tuple(int(value) for value in bundle[size_key])
    if len(size) != 2 or min(size) <= 0:
        raise ValueError(f"Invalid image size: {size}")
    hog = cv2.HOGDescriptor(**bundle["hog_params"])
    if hog.getDescriptorSize() != model.n_features_in_:
        raise ValueError(f"HOG {path}")
    print(f"Loaded {path.name}：{list(labels.values())}")
    return model, size, hog, labels


def classify_region(crop, model_info):
    """HOG→SVM。Resize a gray crop, extract HOG and predict.

    
    Keep the training preprocessing; do not add extra image enhancement.
    """
    model, size, hog, labels = model_info
    interpolation = (
        cv2.INTER_AREA
        if crop.shape[1] >= size[0] and crop.shape[0] >= size[1]
        else cv2.INTER_LINEAR
    )
    image = cv2.resize(crop, size, interpolation=interpolation)
    feature = hog.compute(image).reshape(1, -1).astype(np.float32)
    return labels[int(model.predict(feature)[0])]


def initialize_expression_tracking(frame):
    """Select ROIs and initialize head corners."""
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    head = select_roi(frame, "Select whole head, then Enter")
    regions = []
    for name in ("screen_left_eye", "screen_right_eye", "mouth"):
        rect = select_roi(frame, f"Select {name}, then Enter")
        x, y, w, h = rect
        regions.append({"name": name, "rect": rect,
                        "template": gray[y:y+h, x:x+w].copy()})
    mask = np.zeros(gray.shape, np.uint8)
    x, y, w, h = head
    mask[y:y+h, x:x+w] = 255
    points = cv2.goodFeaturesToTrack(gray, mask=mask, **FEATURE_PARAMS)
    if points is None or len(points) < MIN_POINTS:
        raise ValueError("")
    return gray, points, rect_points(head), regions
```

## 4. Expression recognition and three-axis head rotation


```python
PART_NAMES=('screen_left_eye','screen_right_eye','mouth')

def neutral_row():
    row={'head_tracking_ok':False, 'screen_left_eye':'unknown','screen_right_eye':'unknown','mouth':'unknown'}
    row.update({name+'_crop_valid':False for name in PART_NAMES})
    return row

class ClassicLive:
    def __init__(self, frame, selected=None):
        self.eye_info=load_expression_model(EYE_MODEL_PATH,'eye_size')
        self.mouth_info=load_expression_model(MOUTH_MODEL_PATH,'mouth_size')
        if selected is None:
            self.previous,self.points,self.corners,self.regions=initialize_expression_tracking(frame)
        else:
            # Saved ROIs are for offline checks only
            gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
            x,y,w,h=selected['head'];mask=np.zeros_like(gray);mask[y:y+h,x:x+w]=255
            self.points=cv2.goodFeaturesToTrack(gray,mask=mask,**FEATURE_PARAMS)
            if self.points is None or len(self.points)<MIN_POINTS:raise ValueError('Not enough points')
            self.previous=gray;self.corners=rect_points(selected['head']);self.regions=[]
            for name in PART_NAMES:
                x,y,w,h=selected[name]
                self.regions.append(dict(name=name,rect=(x,y,w,h),template=gray[y:y+h,x:x+w].copy()))
        self.reference=self.points.copy();self.first=True;self.active=True

    def process(self,frame,timestamp):
        gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY);output=frame.copy();row=neutral_row()
        if not self.active:return row,output
        transform=np.eye(3)
        if not self.first:
            current,ref=track_points(self.previous,gray,self.points,self.reference)
            if current is None or len(current)<MIN_POINTS:
                self.active=False;return row,output
            affine,status=cv2.estimateAffinePartial2D(ref.reshape(-1,2),current.reshape(-1,2),method=cv2.RANSAC,ransacReprojThreshold=RANSAC_THRESHOLD)
            if affine is None or status is None or not np.isfinite(affine).all() or status.sum()<MIN_POINTS:
                self.active=False;return row,output
            transform[:2]=affine
            good=status.ravel().astype(bool);self.points=current[good];self.reference=ref[good]
        corners=cv2.perspectiveTransform(self.corners,transform)
        scale=np.linalg.norm(transform[:2,0]);height,width=gray.shape
        if not np.isfinite(corners).all() or not .05<scale<20 or not 100<abs(cv2.contourArea(corners))<width*height:
            self.active=False;return row,output
        self.points,self.reference=replenish(gray,self.points,self.reference,transform,corners)
        inverse=np.linalg.inv(transform)
        aligned=cv2.warpAffine(gray,inverse[:2],(width,height))
        valid_mask=cv2.warpAffine(np.full_like(gray,255),inverse[:2],(width,height),flags=cv2.INTER_NEAREST)
        row['head_tracking_ok']=True
        # Image y points down, VRM y points up
        roll=-float(np.degrees(np.arctan2(transform[1,0],transform[0,0])))
        row['head_pose']={'pitch':0.0,'yaw':0.0,'roll':float(np.clip(roll,-30,30))}

        cv2.polylines(output,[np.rint(corners).astype(np.int32)],True,(255,255,0),1)
        for region in self.regions:
            name=region['name'];rect=region['rect'];method='HEAD_ROI'
            if name!='mouth':
                rect,score,matched=locate_eye(aligned,valid_mask,region)
                method='MATCH' if matched else 'PREDICTED'
            x,y,w,h=rect
            valid=x>=0 and y>=0 and x+w<=width and y+h<=height and np.all(valid_mask[y:y+h,x:x+w]>250)
            if valid:
                row[name]=classify_region(aligned[y:y+h,x:x+w],self.mouth_info if name=='mouth' else self.eye_info)
                row[name+'_crop_valid']=True
            row[name+'_localization']=method
            polygon=cv2.perspectiveTransform(rect_points(rect),transform)
            cv2.polylines(output,[np.rint(polygon).astype(np.int32)],True,(0,255,0),1)
        self.previous=gray;self.first=False
        return row,output

    def close(self):pass



class MediaPipeLive:

    def __init__(self):
        from importlib import import_module
        try:
            mp = import_module('mediapipe')
        except ModuleNotFoundError as exc:
            raise ImportError('MediaPipe is required for this backend. Run `%pip install mediapipe` in this notebook, then restart the kernel.') from exc
        if not FACE_MODEL_PATH.is_file():
            raise FileNotFoundError(FACE_MODEL_PATH)
        self.mp = mp
        self.body_detector = None
        self.body_points = None
        self.body_tick = -1000
        self.pose_pixels = []
        self.show_skeleton = True
        options = mp.tasks.vision.FaceLandmarkerOptions(base_options=mp.tasks.BaseOptions(model_asset_path=str(FACE_MODEL_PATH)), running_mode=mp.tasks.vision.RunningMode.VIDEO, num_faces=1, output_facial_transformation_matrixes=True)
        self.detector = mp.tasks.vision.FaceLandmarker.create_from_options(options)
        self.hand_detector = None
        self.hand_tick = -1000
        self.hands = {}
        self.baseline = np.array([0.28, 0.28])
        self.mouth_baseline = 0.025
        self.head_rotation = None
        self.head_reference = None
        self.raw = None
        self.last = -1
        self.values = np.array([1.0, 1.0, 0.0])
        self.previous_time = None

    def calibrate(self):
        if self.raw is None:
            return False
        self.baseline = np.clip(self.raw[:2], 0.12, 0.5)
        self.mouth_baseline = float(np.clip(self.raw[2], 0, 0.2))
        if self.head_rotation is not None:
            self.head_reference = self.head_rotation.copy()
        return True

    def process(self, frame, timestamp):
        timestamp = max(int(timestamp), self.last + 1)
        self.last = timestamp
        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        result = self.detector.detect_for_video(self.mp.Image(image_format=self.mp.ImageFormat.SRGB, data=rgb), timestamp)
        row = neutral_row()
        preview = frame.copy()
        if not result.face_landmarks:
            self.raw = None
            self.head_rotation = None
            self.values = np.array([1.0, 1.0, 0.0])
            self.previous_time = None
            return (row, preview)
        if result.facial_transformation_matrixes:
            matrix = np.asarray(result.facial_transformation_matrixes[0], dtype=float)[:3, :3]
            u, _, vt = np.linalg.svd(matrix)
            rotation = u @ np.diag([1.0, 1.0, np.linalg.det(u @ vt)]) @ vt
            self.head_rotation = rotation
            if self.head_reference is None:
                self.head_reference = rotation.copy()
            angles = np.asarray(cv2.RQDecomp3x3(rotation @ self.head_reference.T)[0])
            angles = (angles + 180) % 360 - 180
            angles = np.clip(angles, [-30, -50, -30], [30, 50, 30])
            row['head_pose'] = {'pitch': float(angles[0]), 'yaw': float(angles[1]), 'roll': float(angles[2])}
        h, w = frame.shape[:2]
        points = np.array([(p.x * w, p.y * h) for p in result.face_landmarks[0]])
        distance = lambda a, b: np.linalg.norm(points[a] - points[b])
        ratios = []
        centres = []
        for a, b, u, d, u2, d2 in [(33, 133, 159, 145, 158, 153), (362, 263, 386, 374, 385, 380)]:
            ratios.append((distance(u, d) + distance(u2, d2)) / (2 * max(distance(a, b), 1)))
            centres.append((points[a, 0] + points[b, 0]) / 2)
        ratios = np.array(ratios)[np.argsort(centres)]
        mouth = distance(13, 14) / max(distance(61, 291), 1)
        self.raw = np.r_[ratios, mouth]
        target = np.r_[np.clip((ratios / self.baseline - 0.35) / 0.55, 0, 1), np.clip((mouth - self.mouth_baseline - 0.02) / 0.45, 0, 1)]
        dt = (timestamp - self.previous_time) / 1000 if self.previous_time is not None else 1
        self.values += (1 - np.exp(-max(dt, 0.001) / np.array([0.018, 0.018, 0.045]))) * (target - self.values)
        self.previous_time = timestamp
        row.update(head_tracking_ok=True, screen_left_eye='closed' if self.values[0] < 0.4 else 'open', screen_right_eye='closed' if self.values[1] < 0.4 else 'open', mouth='a' if self.values[2] > 0.18 else 'closed')
        row.update({name + '_crop_valid': True for name in PART_NAMES})
        row['continuous'] = {'left_closed': float(1 - self.values[0]), 'right_closed': float(1 - self.values[1]), 'mouth_open': float(self.values[2])}
        return (row, preview)

    def close(self):
        self.detector.close()

class HybridLive:

    def __init__(self, frame, selected=None):
        self.classic = ClassicLive(frame, selected)
        self.pose = None
        try:
            self.pose = MediaPipeLive()
        except Exception:
            self.classic.close()
            raise

    def process(self, frame, timestamp):
        row, preview = self.classic.process(frame, timestamp)
        pose_row, _ = self.pose.process(frame, timestamp)
        row.pop('head_pose', None)
        if pose_row.get('head_pose') and row['head_tracking_ok']:
            row['head_pose'] = pose_row['head_pose']
        return (row, preview)

    def calibrate(self):
        return self.pose.calibrate()

    def close(self):
        self.classic.close()
        if self.pose is not None:
            self.pose.close()
```

## 5. Local state server


```python
STATE_LOCK=Lock()
LIVE_STATE={'row':neutral_row(),'active':False,'mirror':MIRROR,'backend':BACKEND,'sequence':0,'published':0.0}

def publish_state(row,active=False):
    with STATE_LOCK:
        LIVE_STATE.update(row=row,active=bool(active),mirror=MIRROR,backend=BACKEND,
                          sequence=LIVE_STATE['sequence']+1,published=time.monotonic())

def state_snapshot():
    with STATE_LOCK:
        data={k:v for k,v in LIVE_STATE.items() if k!='published'}
        data['age_ms']=(time.monotonic()-LIVE_STATE['published'])*1000 if LIVE_STATE['published'] else 1e9
    return data

class LiveHandler(BaseHTTPRequestHandler):
    def log_message(self,*args):pass
    def do_GET(self):
        path=self.path.split('?',1)[0]
        if path=='/api/state':
            data=json.dumps(state_snapshot(),allow_nan=False).encode('utf-8');kind='application/json'
        else:
            # Explicit allowlist: no directory browsing
            allowed={'/':'live_driver.html','/live_driver.html':'live_driver.html',
                     '/timeline_viseme.mjs':'timeline_viseme.mjs','/live_state.mjs':'live_state.mjs'}
            file=VRM_PATH if path=='/avatar.vrm' else LIVE_DIR/allowed[path] if path in allowed else None
            if file is None or not file.is_file():self.send_error(404);return
            data=file.read_bytes()
            kind='application/javascript' if file.suffix=='.mjs' else 'text/html; charset=utf-8' if file.suffix=='.html' else 'application/octet-stream'
        self.send_response(200);self.send_header('Content-Type',kind);self.send_header('Content-Length',str(len(data)))
        self.send_header('Cache-Control','no-store');self.end_headers()
        try:self.wfile.write(data)
        except (BrokenPipeError,ConnectionResetError,ConnectionAbortedError):pass

def start_live_server(open_browser=True):
    global live_server,live_url
    if 'live_server' in globals():
        live_server.shutdown();live_server.server_close()
    publish_state(neutral_row(),False)
    live_server=ThreadingHTTPServer(('127.0.0.1',0),LiveHandler)
    Thread(target=live_server.serve_forever,daemon=True).start()
    live_url=f'http://127.0.0.1:{live_server.server_port}/live_driver.html'
    print('Live page:',live_url)
    if open_browser:webbrowser.open(live_url)
    return live_url

# Start only a local page; the next cell controls camera access
OPEN_LIVE_PAGE=False  # Set True to open
if OPEN_LIVE_PAGE:start_live_server()
```

## 6. Camera loop


```python
def open_live_camera():
    for backend in (cv2.CAP_DSHOW,cv2.CAP_ANY):
        cap=cv2.VideoCapture(CAMERA_INDEX,backend)
        if cap.isOpened():
            cap.set(cv2.CAP_PROP_FRAME_WIDTH,CAMERA_WIDTH);cap.set(cv2.CAP_PROP_FRAME_HEIGHT,CAMERA_HEIGHT)
            ok,frame=cap.read()
            if ok:return cap,frame
        cap.release()
    raise RuntimeError('Camera unavailable; close other capture apps')

def run_live_camera(limit=0,show=True):
    if BACKEND not in ('classic','hybrid'):raise ValueError('BACKEND must be classic or mediapipe')
    if 'live_server' not in globals():start_live_server()
    cap=None;recognizer=None;count=0;detected=0;started=time.monotonic()
    window='VTuber Head Tilt Live - Q quit | R reset | C calibrate'
    try:
        cap,frame=open_live_camera()
        if MIRROR:frame=cv2.flip(frame,1)
        publish_state(neutral_row(),False)
        recognizer=HybridLive(frame) if BACKEND=='hybrid' else ClassicLive(frame)
        if show:cv2.namedWindow(window,cv2.WINDOW_NORMAL)
        # Start the clock after manual selection
        started=time.monotonic()
        while True:
            tick=time.monotonic()
            row,preview=recognizer.process(frame,int((tick-started)*1000))
            row.update(frame_index=count,time_s=tick-started,processing_ms=(time.monotonic()-tick)*1000)
            publish_state(row,True);count+=1;detected+=int(row['head_tracking_ok'])
            if show:
                text=f"{BACKEND} | L:{row['screen_left_eye']} R:{row['screen_right_eye']} mouth:{row['mouth']} | Q quit"
                cv2.rectangle(preview,(0,0),(preview.shape[1],52),(25,25,25),-1)
                cv2.putText(preview,text,(8,20),cv2.FONT_HERSHEY_SIMPLEX,.42,(200,240,200),1)
                cv2.putText(preview,'R reselect | C calibrate | Q quit',(8,42),cv2.FONT_HERSHEY_SIMPLEX,.42,(220,220,220),1)
                cv2.imshow(window,preview);key=cv2.waitKey(1)&255
                if key==ord('c') and BACKEND=='hybrid':print('Calibrated:',recognizer.calibrate())
                if key in (27,ord('q')) or cv2.getWindowProperty(window,cv2.WND_PROP_VISIBLE)<1:break
                if key==ord('r') and BACKEND in ('classic','hybrid'):
                    publish_state(neutral_row(),False);recognizer.close();recognizer=HybridLive(frame) if BACKEND=='hybrid' else ClassicLive(frame)
            if limit and count>=limit:break
            ok,frame=cap.read()
            if not ok:raise RuntimeError('Camera disconnected')
            if MIRROR:frame=cv2.flip(frame,1)
    finally:
        publish_state(neutral_row(),False)
        if cap is not None:cap.release()
        if recognizer is not None:recognizer.close()
        if show:cv2.destroyAllWindows()
        print({'processed_frames':count,'tracked_frames':detected,'elapsed_seconds':round(time.monotonic()-started,2)})

START_CAMERA=False  # Set True and run this cell
if START_CAMERA:run_live_camera()
```

Set START_LIVE=True and run this cell. The cell remains busy while capture runs. Press Q in the camera window to finish.


```python
# One entry point
START_LIVE = False
if START_LIVE:
    start_live_server(open_browser=True)
    run_live_camera()
```
