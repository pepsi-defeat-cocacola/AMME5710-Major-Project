```python
from pathlib import Path
import csv
import pickle
from datetime import datetime

import cv2
import numpy as np

VIDEO_PATH = Path(
    r"C:\Users\user\Desktop\M2\Computer Vision and Image Processing\Major project\Matt.mp4"
)

MODEL_PATH = Path(
    r"C:\Users\user\Desktop\M2\Computer Vision and Image Processing\Major project"
    r"\mouth_training_runs\20261025_051126_f48f40"
    r"\mouth_hog_svm.pkl"
)

FEATURE_PARAMS = dict(
    maxCorners=150,
    qualityLevel=0.01,
    minDistance=7,
    blockSize=7
)

LK_PARAMS = dict(
    winSize=(21, 21),
    maxLevel=3,
    criteria=(
        cv2.TERM_CRITERIA_EPS | cv2.TERM_CRITERIA_COUNT,
        30,
        0.01
    )
)

FB_THRESHOLD = 1.5
RANSAC_THRESHOLD = 3.0
MIN_POINTS = 10
REPLENISH_THRESHOLD = 80

SEARCH_MARGIN = 0.35
MATCH_THRESHOLD = 0.15

PANEL_WIDTH = 320
WINDOW_NAME = "Mouth prediction | Q quit | Space pause"

# 加载自己训练并保存的模型
with open(MODEL_PATH, "rb") as file:
    bundle = pickle.load(file)

mouth_model = bundle["model"]
mouth_size = tuple(bundle["mouth_size"])
mouth_hog = cv2.HOGDescriptor(**bundle["hog_params"])

id_to_label = {
    int(value): key
    for key, value in bundle["label_to_id"].items()
}

if mouth_hog.getDescriptorSize() != mouth_model.n_features_in_:
    raise ValueError("HOG 特征维数与模型不一致。")

print("模型加载成功。")
print("OpenCV：", cv2.__version__)
print("HOG 特征维数：", mouth_hog.getDescriptorSize())
```

```python
def select_roi(frame, title):
    """人工框选区域。"""

    x, y, w, h = cv2.selectROI(
        title, frame,
        showCrosshair=True,
        fromCenter=False
    )
    cv2.destroyWindow(title)

    if w < 8 or h < 8:
        raise ValueError("区域太小或选择已取消。")

    return int(x), int(y), int(w), int(h)


def rect_points(rect):
    x, y, w, h = rect

    return np.float32([
        [x, y],
        [x + w - 1, y],
        [x + w - 1, y + h - 1],
        [x, y + h - 1]
    ]).reshape(-1, 1, 2)


def initialize(frame):
    """依次选择头部和嘴部。"""

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    head_rect = select_roi(
        frame, "Select whole head, then Enter"
    )

    mask = np.zeros(gray.shape, dtype=np.uint8)
    x, y, w, h = head_rect
    mask[y:y + h, x:x + w] = 255

    points = cv2.goodFeaturesToTrack(
        gray, mask=mask, **FEATURE_PARAMS
    )

    if points is None or len(points) < MIN_POINTS:
        raise ValueError("头部特征点不足，请重新选择。")

    mouths = []

    for name in ("mouth",):
        rect = select_roi(
            frame, f"Select {name}, then Enter"
        )

        x, y, w, h = rect

        mouths.append({
            "name": name,
            "rect": rect,
            "template": gray[y:y + h, x:x + w].copy()
        })

    return gray, points, rect_points(head_rect), mouths


def track_points(previous_gray, current_gray, points, reference):
    """正反向光流检查；reference 保留初始化坐标。"""

    current, status, _ = cv2.calcOpticalFlowPyrLK(
        previous_gray, current_gray,
        points, None, **LK_PARAMS
    )

    if current is None or status is None:
        return None, None

    valid = (
        (status.ravel() == 1)
        & np.isfinite(current.reshape(-1, 2)).all(axis=1)
    )

    if not valid.any():
        return None, None

    old = points[valid]
    current = current[valid]
    reference = reference[valid]

    backward, status_back, _ = cv2.calcOpticalFlowPyrLK(
        current_gray, previous_gray,
        current, None, **LK_PARAMS
    )

    if backward is None or status_back is None:
        return None, None

    error = np.linalg.norm(
        old.reshape(-1, 2) - backward.reshape(-1, 2),
        axis=1
    )

    height, width = current_gray.shape
    xy = current.reshape(-1, 2)

    reliable = (
        (status_back.ravel() == 1)
        & np.isfinite(error)
        & (error < FB_THRESHOLD)
        & (xy[:, 0] >= 0)
        & (xy[:, 0] < width)
        & (xy[:, 1] >= 0)
        & (xy[:, 1] < height)
    )

    return current[reliable], reference[reliable]


def replenish(gray, points, reference, transform, corners):
    """为整体跟踪补点，不给普通角点赋予五官语义。"""

    if len(points) >= REPLENISH_THRESHOLD:
        return points, reference

    mask = np.zeros(gray.shape, dtype=np.uint8)

    cv2.fillConvexPoly(
        mask,
        np.rint(corners).astype(np.int32).reshape(-1, 2),
        255
    )

    mask = cv2.erode(
        mask, np.ones((9, 9), dtype=np.uint8)
    )

    for point in points.reshape(-1, 2):
        cv2.circle(
            mask,
            tuple(np.rint(point).astype(int)),
            10, 0, -1
        )

    params = FEATURE_PARAMS.copy()
    params["maxCorners"] = (
        FEATURE_PARAMS["maxCorners"] - len(points)
    )

    if params["maxCorners"] <= 0:
        return points, reference

    new_points = cv2.goodFeaturesToTrack(
        gray, mask=mask, **params
    )

    if new_points is None:
        return points, reference

    new_reference = cv2.perspectiveTransform(
        new_points, np.linalg.inv(transform)
    )

    valid = np.isfinite(
        new_reference.reshape(-1, 2)
    ).all(axis=1)

    return (
        np.concatenate([points, new_points[valid]], axis=0),
        np.concatenate([reference, new_reference[valid]], axis=0)
    )


def locate_mouth(aligned, valid_mask, mouth):
    """局部模板匹配，失败时退回整体运动预测。"""

    x, y, w, h = mouth["rect"]
    height, width = aligned.shape

    mx = max(4, int(w * SEARCH_MARGIN))
    my = max(4, int(h * SEARCH_MARGIN))

    x0, y0 = max(0, x - mx), max(0, y - my)
    x1 = min(width, x + w + mx)
    y1 = min(height, y + h + my)

    search = aligned[y0:y1, x0:x1]
    rect = mouth["rect"]
    score = float("inf")
    matched = False

    if search.shape[0] >= h and search.shape[1] >= w:
        response = cv2.matchTemplate(
            search,
            mouth["template"],
            cv2.TM_SQDIFF_NORMED
        )

        valid_search = (
            valid_mask[y0:y1, x0:x1] > 250
        ).astype(np.float32)

        coverage = cv2.matchTemplate(
            valid_search,
            np.ones((h, w), dtype=np.float32),
            cv2.TM_CCORR
        ) / float(w * h)

        response[coverage < 0.999] = np.inf
        response[~np.isfinite(response)] = np.inf

        if np.isfinite(response).any():
            score, _, location, _ = cv2.minMaxLoc(response)

            if score <= MATCH_THRESHOLD:
                rect = (
                    x0 + location[0],
                    y0 + location[1],
                    w, h
                )
                matched = True

    return rect, float(score), matched


def predict_mouth(crop):
    """与嘴部训练单元格 7 使用相同的预处理。"""

    interpolation = (
        cv2.INTER_AREA
        if crop.shape[1] >= mouth_size[0]
        and crop.shape[0] >= mouth_size[1]
        else cv2.INTER_LINEAR
    )

    resized = cv2.resize(
        crop, mouth_size,
        interpolation=interpolation
    )

    features = mouth_hog.compute(resized).reshape(
        1, -1
    ).astype(np.float32)

    predicted_id = int(mouth_model.predict(features)[0])

    return id_to_label[predicted_id]


def draw_preview(panel, crop, index):
    """等比例显示嘴部裁剪图，保留 index 参数。"""

    row_height = panel.shape[0]
    top = index * row_height + 85
    available_height = row_height - 95

    if available_height <= 0:
        return

    scale = min(
        (PANEL_WIDTH - 20) / crop.shape[1],
        available_height / crop.shape[0]
    )

    size = (
        max(1, int(crop.shape[1] * scale)),
        max(1, int(crop.shape[0] * scale))
    )

    preview = cv2.resize(crop, size)
    preview = cv2.cvtColor(preview, cv2.COLOR_GRAY2BGR)

    panel[
        top:top + size[1],
        10:10 + size[0]
    ] = preview
```

```python
SAVE_RESULTS = (
    input("是否保存嘴部结果视频和逐帧 CSV？输入 y 保存：")
    .strip().lower() == "y"
)

cap = cv2.VideoCapture(str(VIDEO_PATH))
writer = None
csv_file = None
saved_frames = 0

try:
    if not cap.isOpened():
        raise FileNotFoundError(f"无法打开：{VIDEO_PATH}")

    success, first_frame = cap.read()

    if not success:
        raise RuntimeError("无法读取第一帧。")

    height, width = first_frame.shape[:2]

    if height < 240:
        raise ValueError("视频高度太小，请使用较高分辨率的视频。")

    fps = cap.get(cv2.CAP_PROP_FPS)

    if not np.isfinite(fps) or fps <= 0:
        fps = 30.0

    delay = max(1, int(round(1000 / fps)))

    previous_gray, points, head_corners, mouths = initialize(
        first_frame
    )

    reference = points.copy()

    if SAVE_RESULTS:
        run_id = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
        output_dir = (
            VIDEO_PATH.parent
            / "mouth_prediction_runs"
            / run_id
        )

        output_dir.mkdir(parents=True, exist_ok=False)

        video_path = output_dir / "mouth_prediction.mp4"
        csv_path = output_dir / "mouth_predictions.csv"

        writer = cv2.VideoWriter(
            str(video_path),
            cv2.VideoWriter_fourcc(*"mp4v"),
            fps,
            (width + PANEL_WIDTH, height)
        )

        if not writer.isOpened():
            raise RuntimeError("无法创建输出视频。")

        csv_file = open(
            csv_path, "w",
            newline="", encoding="utf-8-sig"
        )

        fields = [
            "frame_index", "time_s", "mouth",
            "head_tracking_ok", "localization",
            "template_ssd", "crop_valid", "prediction"
        ]

        csv_writer = csv.DictWriter(
            csv_file, fieldnames=fields
        )
        csv_writer.writeheader()

    cv2.namedWindow(WINDOW_NAME, cv2.WINDOW_NORMAL)

    frame = first_frame
    frame_index = 0
    tracking_active = True

    while True:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        output = frame.copy()
        panel = np.zeros(
            (height, PANEL_WIDTH, 3),
            dtype=np.uint8
        )

        transform = None

        if frame_index == 0:
            transform = np.eye(3, dtype=np.float64)

        elif tracking_active:
            current, matched_reference = track_points(
                previous_gray, gray, points, reference
            )

            if current is None or len(current) < MIN_POINTS:
                tracking_active = False

            else:
                affine, status = cv2.estimateAffinePartial2D(
                    matched_reference.reshape(-1, 2),
                    current.reshape(-1, 2),
                    method=cv2.RANSAC,
                    ransacReprojThreshold=RANSAC_THRESHOLD,
                    maxIters=2000,
                    confidence=0.99,
                    refineIters=10
                )

                if (
                    affine is None
                    or status is None
                    or not np.isfinite(affine).all()
                ):
                    tracking_active = False

                else:
                    inliers = status.ravel().astype(bool)

                    if np.count_nonzero(inliers) < MIN_POINTS:
                        tracking_active = False

                    else:
                        transform = np.vstack([
                            affine, [0.0, 0.0, 1.0]
                        ])

                        points = current[inliers].copy()
                        reference = matched_reference[inliers].copy()

        if tracking_active:
            scale = np.linalg.norm(transform[:2, 0])

            if not 0.05 < scale < 20:
                tracking_active = False

        if tracking_active:
            corners = cv2.perspectiveTransform(
                head_corners, transform
            )

            xy = corners.reshape(-1, 2)
            valid_geometry = np.isfinite(xy).all()

            if valid_geometry:
                valid_geometry = (
                    (np.abs(xy[:, 0]) < 3 * width).all()
                    and (np.abs(xy[:, 1]) < 3 * height).all()
                )

            if valid_geometry:
                area = abs(cv2.contourArea(xy.astype(np.float32)))
                valid_geometry = 100 < area < width * height

            if not valid_geometry:
                tracking_active = False

        if tracking_active:
            points, reference = replenish(
                gray, points, reference, transform, corners
            )

            inverse = np.linalg.inv(transform)

            aligned = cv2.warpAffine(
                gray, inverse[:2], (width, height),
                flags=cv2.INTER_LINEAR,
                borderMode=cv2.BORDER_CONSTANT,
                borderValue=0
            )

            valid_mask = cv2.warpAffine(
                np.full(gray.shape, 255, dtype=np.uint8),
                inverse[:2], (width, height),
                flags=cv2.INTER_NEAREST,
                borderMode=cv2.BORDER_CONSTANT,
                borderValue=0
            )

            cv2.polylines(
                output,
                [np.rint(corners).astype(np.int32)],
                True, (255, 255, 0), 2
            )

        for index, mouth in enumerate(mouths):
            prediction = "unknown"
            localization = "lost"
            score = None
            crop_valid = False
            crop = None

            if tracking_active:
                rect, score, matched = locate_mouth(
                    aligned, valid_mask, mouth
                )

                localization = (
                    "MATCH" if matched else "PREDICTED"
                )

                x, y, w, h = rect
                crop = aligned[y:y + h, x:x + w]
                crop_mask = valid_mask[y:y + h, x:x + w]

                crop_valid = (
                    crop.shape == (h, w)
                    and crop_mask.shape == (h, w)
                    and crop.size > 0
                    and np.all(crop_mask > 250)
                )

                # 即使模板失配，只要预测裁剪有效，也执行嘴部分类
                if crop_valid:
                    prediction = predict_mouth(crop)

                color = (
                    (0, 255, 0) if prediction == "open"
                    else (0, 0, 255) if prediction == "closed"
                    else (0, 165, 255)
                )

                polygon = cv2.perspectiveTransform(
                    rect_points(rect), transform
                )

                cv2.polylines(
                    output,
                    [np.rint(polygon).astype(np.int32)],
                    True, color, 2
                )

            # 只有一个嘴部区域，使用整个面板高度
            row_top = index * height

            color = (
                (0, 255, 0) if prediction == "open"
                else (0, 0, 255) if prediction == "closed"
                else (0, 165, 255)
            )

            cv2.putText(
                panel, mouth["name"],
                (10, row_top + 23),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5, (255, 255, 255), 1
            )

            cv2.putText(
                panel, f"State: {prediction.upper()}",
                (10, row_top + 48),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65, color, 2
            )

            score_text = (
                f"{score:.3f}"
                if score is not None and np.isfinite(score)
                else "invalid"
            )

            cv2.putText(
                panel,
                f"{localization} | SSD={score_text}",
                (10, row_top + 72),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.45, (255, 255, 255), 1
            )

            if crop_valid:
                draw_preview(panel, crop, index)

            if csv_file is not None:
                csv_writer.writerow({
                    "frame_index": frame_index,
                    "time_s": frame_index / fps,
                    "mouth": mouth["name"],
                    "head_tracking_ok": tracking_active,
                    "localization": localization,
                    "template_ssd": (
                        score
                        if score is not None and np.isfinite(score)
                        else ""
                    ),
                    "crop_valid": bool(crop_valid),
                    "prediction": prediction
                })

        if not tracking_active:
            cv2.putText(
                output,
                "Tracking lost. Rerun this cell to initialize again.",
                (15, 30),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.55, (0, 0, 255), 2
            )

        combined = np.hstack([output, panel])

        if writer is not None:
            writer.write(combined)
            saved_frames += 1

        cv2.imshow(WINDOW_NAME, combined)
        key = cv2.waitKey(delay) & 0xFF

        if key == ord(" "):
            while True:
                key = cv2.waitKey(30) & 0xFF

                if key in (ord(" "), ord("q"), 27):
                    break

                if cv2.getWindowProperty(
                    WINDOW_NAME, cv2.WND_PROP_VISIBLE
                ) < 1:
                    key = ord("q")
                    break

        if key in (ord("q"), 27):
            break

        if cv2.getWindowProperty(
            WINDOW_NAME, cv2.WND_PROP_VISIBLE
        ) < 1:
            break

        previous_gray = gray.copy()

        success, frame = cap.read()

        if not success:
            print("嘴部视频处理结束。")
            break

        frame_index += 1

finally:
    cap.release()

    if writer is not None:
        writer.release()

    if csv_file is not None:
        csv_file.close()

    cv2.destroyAllWindows()

if SAVE_RESULTS:
    print(f"保存的嘴部视频帧数：{saved_frames}")
    print("保存目录：", output_dir)
```
