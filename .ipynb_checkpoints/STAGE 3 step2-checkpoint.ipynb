```python
from pathlib import Path, PureWindowsPath
from collections import Counter
import csv
import json
import pickle

import cv2
import numpy as np

from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    precision_recall_fscore_support
)

# 修改为所有标注数据所在的父文件夹
DATASET_ROOT = Path(
    r"C:\Users\user\Desktop\M2\Computer Vision and Image Processing\Major project\mouth_dataset"
)

# 第一次运行保持为空
# 代码会列出原视频名称，随后在这里填写人物编号
PERSON_BY_VIDEO = {
    "train01_asian_female_12s.mp4": "p01",
    "Elon Musk.mp4": "p02",
    "Ishihara Satomi.mp4": "p03",
    "train04_white_female_10s.mp4": "p04",
    "Zelenskyy.mp4": "p05",
    "train05_elderly_white_female_9s.mp4": "p06",
    "train06a_asian_male_6s.mp4": "p07",
    "train08_black_male_suit_10s.mp4": "p08",
    "train09_black_male_glasses_11s.mp4": "p09",
    # 示例：使用下面打印出的完整文件名，包含扩展名
    # "test2.mp4": "p01",
    # "person2.mp4": "p02",
    # "person3.mp4": "p03",
    # "person4.mp4": "p04",
    # "person5.mp4": "p05",
}

# 先固定划分，再训练
TRAIN_PEOPLE = {"p01", "p02", "p03", "p06", "p07", "p08", "p09"}
VAL_PEOPLE = {"p04"}
TEST_PEOPLE = {"p05"}

LABEL_TO_ID = {"open": 0, "closed": 1}

records = []
seen = {}
duplicate_count = 0

csv_paths = sorted(DATASET_ROOT.rglob("labels.csv"))

if not csv_paths:
    raise FileNotFoundError(
        f"没有找到 labels.csv：{DATASET_ROOT}"
    )

for csv_path in csv_paths:
    with open(csv_path, newline="", encoding="utf-8-sig") as file:
        for row in csv.DictReader(file):
            label = row["label"].strip().lower()

            if label not in LABEL_TO_ID:
                raise ValueError(f"未知标签：{label}，来源：{csv_path}")

            # 原采集代码记录的是 Windows 视频路径
            video_name = PureWindowsPath(
                row["source_video"].replace("/", "\\")
            ).name

            # 相同视频、帧、左右眼只保留一次
            key = (
                video_name,
                int(row["frame_index"]),
                row["mouth"]
            )

            if key in seen:
                if seen[key] != label:
                    raise ValueError(
                        f"重复样本标签冲突，请检查：{key}"
                    )
                duplicate_count += 1
                continue

            image_path = csv_path.parent / row["filename"]

            if not image_path.is_file():
                raise FileNotFoundError(image_path)

            seen[key] = label

            records.append({
                "image_path": image_path,
                "video": video_name,
                "frame_index": int(row["frame_index"]),
                "mouth": row["mouth"],
                "label": label,
                "y": LABEL_TO_ID[label]
            })

video_names = sorted({record["video"] for record in records})

print("发现的原视频及样本数量：")
for name in video_names:
    counts = Counter(
        record["label"]
        for record in records
        if record["video"] == name
    )
    print(
        f"{name}: open={counts['open']}, closed={counts['closed']}"
    )

print("已忽略重复样本：", duplicate_count)

missing_names = [
    name for name in video_names
    if name not in PERSON_BY_VIDEO
]

if missing_names:
    raise ValueError(
        "请根据上面的视频文件名填写 PERSON_BY_VIDEO，"
        "然后重新运行本单元格。"
    )

# 一个人有多段视频时，全部映射到相同人物编号
for record in records:
    record["person"] = PERSON_BY_VIDEO[record["video"]]

if (
    TRAIN_PEOPLE & VAL_PEOPLE
    or TRAIN_PEOPLE & TEST_PEOPLE
    or VAL_PEOPLE & TEST_PEOPLE
):
    raise ValueError("训练、验证和测试人物不能重叠。")

expected_people = TRAIN_PEOPLE | VAL_PEOPLE | TEST_PEOPLE
actual_people = {record["person"] for record in records}

if actual_people != expected_people:
    raise ValueError(
        f"人物编号不一致。实际：{actual_people}；"
        f"划分设置：{expected_people}"
    )

for split_name, people in [
    ("train", TRAIN_PEOPLE),
    ("validation", VAL_PEOPLE),
    ("test", TEST_PEOPLE)
]:
    selected = [
        record for record in records
        if record["person"] in people
    ]

    counts = Counter(record["label"] for record in selected)

    print(
        f"{split_name}: people={sorted(people)}, "
        f"open={counts['open']}, closed={counts['closed']}"
    )

    if counts["open"] == 0 or counts["closed"] == 0:
        raise ValueError(
            f"{split_name} 缺少睁眼或闭眼样本，请补充标注。"
        )
```

```python
# OpenCV 的尺寸顺序为 (宽, 高)
MOUTH_SIZE = (64, 32)

HOG_PARAMS = {
    "_winSize": MOUTH_SIZE,
    "_blockSize": (16, 16),
    "_blockStride": (8, 8),
    "_cellSize": (8, 8),
    "_nbins": 9
}

hog = cv2.HOGDescriptor(**HOG_PARAMS)


def extract_mouth_hog(image_path):
    """读取原始眼睛裁剪图并提取固定长度的 HOG。"""

    image_bytes = np.fromfile(
        str(image_path),
        dtype=np.uint8
    )

    image = cv2.imdecode(
        image_bytes,
        cv2.IMREAD_GRAYSCALE
    )

    if image is None:
        raise RuntimeError(f"无法读取图片：{image_path}")

    interpolation = (
        cv2.INTER_AREA
        if image.shape[1] >= MOUTH_SIZE[0]
        and image.shape[0] >= MOUTH_SIZE[1]
        else cv2.INTER_LINEAR
    )

    resized = cv2.resize(
        image,
        MOUTH_SIZE,
        interpolation=interpolation
    )

    return hog.compute(resized).ravel()


def build_features(selected_records):
    X = np.stack([
        extract_mouth_hog(record["image_path"])
        for record in selected_records
    ]).astype(np.float32)

    y = np.array([
        record["y"] for record in selected_records
    ], dtype=np.int32)

    return X, y


train_records = [
    record for record in records
    if record["person"] in TRAIN_PEOPLE
]

val_records = [
    record for record in records
    if record["person"] in VAL_PEOPLE
]

test_records = [
    record for record in records
    if record["person"] in TEST_PEOPLE
]

X_train, y_train = build_features(train_records)
X_val, y_val = build_features(val_records)

print("训练特征矩阵：", X_train.shape)
print("验证特征矩阵：", X_val.shape)

best_model = None
best_params = None
best_score = -1.0
validation_results = []

for kernel in ("linear", "rbf"):
    for C in (0.1, 1.0, 10.0):
        for class_weight in (None, "balanced"):
            params = {
                "kernel": kernel,
                "C": C,
                "gamma": "scale",
                "class_weight": class_weight
            }

            model = SVC(**params)
            model.fit(X_train, y_train)

            prediction = model.predict(X_val)

            precision, recall, f1, support = (
                precision_recall_fscore_support(
                    y_val,
                    prediction,
                    labels=[0, 1],
                    zero_division=0
                )
            )

            macro_f1 = float(np.mean(f1))

            result = {
                **params,
                "val_accuracy": float(
                    accuracy_score(y_val, prediction)
                ),
                "val_macro_f1": macro_f1,
                "val_closed_precision": float(precision[1]),
                "val_closed_recall": float(recall[1]),
                "val_closed_f1": float(f1[1])
            }

            validation_results.append(result)

            print(
                f"kernel={kernel:6s}, C={C:4.1f}, "
                f"weight={str(class_weight):8s} | "
                f"macro-F1={macro_f1:.3f}, "
                f"closed recall={recall[1]:.3f}, "
                f"closed F1={f1[1]:.3f}"
            )

            # 分数相同时，保留先测试的配置
            if macro_f1 > best_score:
                best_score = macro_f1
                best_params = params.copy()
                best_model = model

print("\n最佳参数：", best_params)
print("最佳验证 macro-F1：", best_score)

print("\n最佳配置的验证集结果：")
print(classification_report(
    y_val,
    best_model.predict(X_val),
    labels=[0, 1],
    target_names=["open", "closed"],
    digits=4,
    zero_division=0
))
```

```python
# 参数已确定，将训练集和验证集合并用于最终训练
X_fit = np.concatenate([X_train, X_val], axis=0)
y_fit = np.concatenate([y_train, y_val], axis=0)

final_model = SVC(**best_params)
final_model.fit(X_fit, y_fit)

# 到这里才提取最终测试集特征并评价
X_test, y_test = build_features(test_records)
test_prediction = final_model.predict(X_test)

test_report = classification_report(
    y_test,
    test_prediction,
    labels=[0, 1],
    target_names=["open", "closed"],
    digits=4,
    zero_division=0
)

test_report_dict = classification_report(
    y_test,
    test_prediction,
    labels=[0, 1],
    target_names=["open", "closed"],
    output_dict=True,
    zero_division=0
)

test_matrix = confusion_matrix(
    y_test,
    test_prediction,
    labels=[0, 1]
)

# 全部预测为睁眼的基线
baseline_prediction = np.zeros_like(y_test)
baseline_accuracy = accuracy_score(
    y_test,
    baseline_prediction
)

print("最终测试人物：", sorted(TEST_PEOPLE))
print("\n全部预测张嘴的准确率：", baseline_accuracy)
print("\nHOG + SVM 测试结果：")
print(test_report)

print("混淆矩阵：行是真实类别，列是预测类别")
print("类别顺序：[open, closed]")
print(test_matrix)

# 每次保存到新目录，保留各轮实验
from datetime import datetime
import uuid

run_name = (
    datetime.now().strftime("%Y%m%d_%H%M%S")
    + "_"
    + uuid.uuid4().hex[:6]
)

RUN_DIR = DATASET_ROOT.parent / "mouth_training_runs" / run_name
RUN_DIR.mkdir(parents=True, exist_ok=False)

model_bundle = {
    "model": final_model,
    "hog_params": HOG_PARAMS,
    "mouth_size": MOUTH_SIZE,
    "label_to_id": LABEL_TO_ID,
    "best_params": best_params,
    "person_by_video": PERSON_BY_VIDEO,
    "training_people": sorted(TRAIN_PEOPLE | VAL_PEOPLE),
    "test_people": sorted(TEST_PEOPLE),
    "preprocessing": "grayscale, resize only; no equalization"
}

with open(RUN_DIR / "mouth_hog_svm.pkl", "wb") as file:
    pickle.dump(model_bundle, file)

with open(
    RUN_DIR / "validation_results.csv",
    "w",
    newline="",
    encoding="utf-8-sig"
) as file:
    writer = csv.DictWriter(
        file,
        fieldnames=list(validation_results[0].keys())
    )
    writer.writeheader()
    writer.writerows(validation_results)

# 记录实际使用的样本和划分
with open(
    RUN_DIR / "sample_manifest.csv",
    "w",
    newline="",
    encoding="utf-8-sig"
) as file:
    fieldnames = [
        "image_path", "video", "frame_index",
        "mouth", "person", "label", "split"
    ]

    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()

    for record in records:
        split = (
            "train"
            if record["person"] in TRAIN_PEOPLE
            else "validation"
            if record["person"] in VAL_PEOPLE
            else "test"
        )

        writer.writerow({
            key: (
                str(record["image_path"])
                if key == "image_path"
                else split
                if key == "split"
                else record[key]
            )
            for key in fieldnames
        })

with open(
    RUN_DIR / "test_metrics.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        {
            "best_params": best_params,
            "test_report": test_report_dict,
            "confusion_matrix": test_matrix.tolist(),
            "always_open_accuracy": float(baseline_accuracy)
        },
        file,
        ensure_ascii=False,
        indent=2
    )

with open(
    RUN_DIR / "test_report.txt",
    "w",
    encoding="utf-8"
) as file:
    file.write(test_report)
    file.write("\nConfusion matrix [open, closed]:\n")
    file.write(str(test_matrix))
    file.write(
        f"\nAlways-open accuracy: {baseline_accuracy:.6f}\n"
    )

print("\n模型和实验记录已保存到：")
print(RUN_DIR)
```
