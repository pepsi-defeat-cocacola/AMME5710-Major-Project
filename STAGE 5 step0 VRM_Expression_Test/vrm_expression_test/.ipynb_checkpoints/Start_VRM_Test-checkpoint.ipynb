先将此 notebook 与 start_viewer.py、index.html 放在同一解压目录。运行下一格后保持运行；浏览器中选择本地 VRM。中断 kernel 可停止服务。

```python
from pathlib import Path
from functools import partial
from http.server import SimpleHTTPRequestHandler, ThreadingHTTPServer
from threading import Thread

from IPython.display import IFrame, display


# index.html 与 notebook 放在同一个文件夹
VRM_PREVIEW_DIR = Path.cwd().resolve()

if not (VRM_PREVIEW_DIR / "index.html").is_file():
    raise FileNotFoundError(
        f"没有找到：{VRM_PREVIEW_DIR / 'index.html'}\n"
        "请确认 notebook 的工作目录包含 index.html。"
    )


# 重复运行本单元格时，先关闭之前启动的服务
if "vrm_preview_server" in globals():
    vrm_preview_server.shutdown()
    vrm_preview_server.server_close()


# 只向本机提供该文件夹中的页面
handler = partial(
    SimpleHTTPRequestHandler,
    directory=str(VRM_PREVIEW_DIR)
)

# 端口设为 0，由系统选择空闲端口
vrm_preview_server = ThreadingHTTPServer(
    ("127.0.0.1", 0),
    handler
)

# 后台运行，因此本单元格可以正常执行结束
vrm_preview_thread = Thread(
    target=vrm_preview_server.serve_forever,
    daemon=True
)
vrm_preview_thread.start()

vrm_preview_url = (
    f"http://127.0.0.1:"
    f"{vrm_preview_server.server_port}/index.html"
)

print("预览地址：", vrm_preview_url)
print("在下面的页面选择本地 VRM 文件。")
print("如果嵌入页面显示异常，可以在浏览器新标签页打开上述地址。")

display(IFrame(
    src=vrm_preview_url,
    width="100%",
    height=800
))
```

```python
if "vrm_preview_server" in globals():
    vrm_preview_server.shutdown()
    vrm_preview_server.server_close()
    del vrm_preview_server
    print("VRM 预览服务已停止。")
```
