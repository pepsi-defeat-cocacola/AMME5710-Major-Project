Keep this notebook, start_viewer.py and index.html in the same extracted folder. Run the next cell, keep the kernel running and select a local VRM file in the browser. Interrupt the kernel to stop the server.

```python
from pathlib import Path
from functools import partial
from http.server import SimpleHTTPRequestHandler, ThreadingHTTPServer
from threading import Thread

from IPython.display import IFrame, display


# Keep index.html in the same folder as this notebook
VRM_PREVIEW_DIR = Path.cwd().resolve()

if not (VRM_PREVIEW_DIR / "index.html").is_file():
    raise FileNotFoundError(
        f"File not found: {VRM_PREVIEW_DIR / 'index.html'}\n"
        "Make sure the notebook working folder contains index.html."
    )


# Stop the previous server before rerunning this cell
if "vrm_preview_server" in globals():
    vrm_preview_server.shutdown()
    vrm_preview_server.server_close()


# Serve this folder on this computer only
handler = partial(
    SimpleHTTPRequestHandler,
    directory=str(VRM_PREVIEW_DIR)
)

# Use port 0 so the system picks a free port
vrm_preview_server = ThreadingHTTPServer(
    ("127.0.0.1", 0),
    handler
)

# Run the server in the background so this cell can finish
vrm_preview_thread = Thread(
    target=vrm_preview_server.serve_forever,
    daemon=True
)
vrm_preview_thread.start()

vrm_preview_url = (
    f"http://127.0.0.1:"
    f"{vrm_preview_server.server_port}/index.html"
)

print("Preview URL:", vrm_preview_url)
print("Select a local VRM file on the page below.")
print("If the preview does not display correctly, open the URL above in a new tab.")

display(IFrame(
    src=vrm_preview_url,
    width="100%",
    height=800
))
```

```python
if "vrm_preview_server" in globals():
    vrm_preview_server.shutdown()
    vrm_preview_server.server_close()
    del vrm_preview_server
    print("VRM preview server stopped.")
```
